```python
import pandas as pd
goats=pd.read_csv("Goats.csv")
goats["DOB/P. D"]=pd.to_datetime(goats["DOB/P. D"],format="%d-%b-%y")
end_time=pd.to_datetime("19-may-26",format="%d-%b-%y")
goats["Total Days"]=(end_time-goats["DOB/P. D"]).dt.days
Feeding_Cost=pd.read_csv("Feeding_Cost.csv")
Health_Record=pd.read_csv("Health_Record.csv")
data=pd.merge(goats,Health_Record, on="ID",how="inner")
data=pd.merge(data,Feeding_Cost,on="ID",how="inner")
data["Total Cost"]=data["Daily_Cost"]*data["Total Days"]
data["Kids"]=[0,2,0,2,2,2,0,1,0,0]
data["Kids Age"]=[0,6,0,2,0.5,1,0,4,0,0]
data[["Last Vaccination","Deworming_Date"]]=data[["Last Vaccination","Deworming_Date"]].fillna("Unknown")

data["Valuation"]=(data["Kids"]*data["Kids Age"]*3000)-(data["Total Cost"])
print("Top Valuation Goat\n",data.loc[data["Valuation"].idxmax()])
print("Low Valuation Goat\n",data.loc[data["Valuation"].idxmin()])
print(data.groupby("Gender")["Valuation"].sum())
print(data.groupby("Gender")["Valuation"].mean())
print(data.sort_values("Valuation",ascending=False))
```
